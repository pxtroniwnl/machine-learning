# ¿Que es el Machine Learning?

En términos sencillos, el Machine Learning (Aprendizaje Automático) es una rama de la Inteligencia Artificial que permite que las computadoras aprendan y tomen decisiones a partir de datos, sin necesidad de ser programadas explícitamente para cada tarea.


## ¿Como funciona?

A diferencia de la programación tradicional, donde tú le das las reglas al computador (ej. "Si pasa A, haz B"), en el Machine Learning tú le entregas datos y las respuestas esperadas, y el algoritmo encuentra el patrón o la regla por sí mismo

Existen como 3 pilares fundamentales:

1. Aprendizaje Supervisado: El modelo aprende con datos etiquetados (sabemos la respuesta de antemano). Es lo que aplicarás mucho en tu curso de Modelos de Regresión, donde buscas predecir un valor numérico basándote en variables históricas.
2. Aprendizaje No Supervisado: El modelo busca patrones o grupos ocultos en datos que no tienen etiquetas (ej. agrupar clientes por comportamiento de compra).
3. Aprendizaje por Refuerzo: El algoritmo aprende mediante ensayo y error, recibiendo "premios" o "castigos" según sus acciones (muy común en robótica o videojuegos).
