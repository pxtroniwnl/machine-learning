Los Arboles de Decision son uno de los Algoritmos de Machine Learning mas conocidos y que mejor funcionan para cualquier tarea tanto de clasificacion como de regresion y son los **componentes** principales de los *Random Forests*

Vemos en el Dataset de Iris como se ve un arbol de decision de clasificacion

```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier

iris = load_iris()
X = iris.data[:, 2:] # petal length and width
y = iris.target
tree_clf = DecisionTreeClassifier(max_depth=2)
tree_clf.fit(X, y)
```

Y ademas podemos ver el arbol de decision que fue entrenado usando:

```python
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree

plt.figure(figsize=(15, 10))
plot_tree(tree_clf, 
          feature_names=iris.feature_names,  
          class_names=iris.target_names, 
          filled=True, 
          rounded=True, 
          fontsize=12)

plt.title("Visualización del Árbol de Decisión (Iris Dataset)")
plt.show()
```

Veamos como funciona haciendo predicciones. Supon que tu quieres encontrar  las Flores Iris y tu quieres clasificarlo. Siempre se inicia en el nodo *raiz* con una *depth = 0* y si este nodo es mas pequeño que 2.45 cm si este lo es, se va al hijo izquierda y asi sucesivamente. Ahora suponte que encuentras otra flor y la longitud de esta es mas grande de 2.45cm tu debes entonces moverte al hijo derecho el cual no es un nodo hoja asi que hace otra pregunta y es que si el *width <= 1.75* si es asi se va al nodo izquierdo el cual es versicolor sino al derecho que es Virginica y asi sucedivamente siempre llegando a nodos hojas es muy simple.
